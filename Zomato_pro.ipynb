# Project Summary – Zomato Restaurant Clustering

The **Zomato Restaurant Clustering** project focuses on analyzing restaurant-related data to identify meaningful patterns and group restaurants based on their characteristics, customer ratings, popularity, pricing, and cuisine diversity. The main objective of this project is to apply data analytics, statistical techniques, natural language processing (NLP), and unsupervised machine learning algorithms to understand the restaurant industry and discover useful business insights.

The project uses two datasets: **Zomato Restaurant Names and Metadata** and **Zomato Restaurant Reviews**. The restaurant metadata dataset contains information such as restaurant names, costs, collections, cuisines, and operating timings. The reviews dataset includes restaurant names, reviewer details, customer reviews, ratings, timestamps, and pictures. Both datasets were explored and analyzed to understand their structure, identify missing values, detect duplicate records, and examine data types.

The initial phase involved **data cleaning and preprocessing**. Duplicate records were identified and removed, missing categorical values were handled using suitable placeholder values, and numerical and date-related columns were converted into appropriate data types. Missing review text, ratings, and timestamps were retained where appropriate instead of replacing them with arbitrary values. Restaurant names were standardized to support accurate merging of the two datasets.

In the **Exploratory Data Analysis (EDA)** phase, different visualizations were created to understand the distribution of restaurant costs and ratings, popular cuisines, customer engagement, and relationships between different variables. Bar charts, histograms, box plots, scatter plots, heatmaps, and pair plots were used to identify patterns and trends. These visualizations helped examine relationships between restaurant pricing, average ratings, review counts, cuisine diversity, and customer picture uploads.

The project also included **hypothesis testing** to investigate relationships and differences within the restaurant data. Statistical tests such as the independent two-sample Welch's t-test and Pearson correlation test were used to evaluate research hypotheses. P-values were analyzed using a significance level of 0.05 to determine whether sufficient statistical evidence existed to reject the null hypotheses.

During **feature engineering and preprocessing**, important features were extracted and prepared for machine learning. These included restaurant cost, average rating, review count, cuisine count, and average number of pictures. Cuisine information was converted into numerical features using Multi-Hot Encoding. Customer reviews were also processed using NLP techniques such as contraction expansion, lowercasing, punctuation removal, stopword removal while preserving negation words, tokenization, lemmatization, and TF-IDF vectorization. Numerical features were transformed using logarithmic transformations where appropriate and standardized using StandardScaler.

For the machine learning implementation, three unsupervised clustering algorithms were explored: **K-Means Clustering, Agglomerative Hierarchical Clustering, and DBSCAN**. Different parameter combinations were evaluated to identify suitable clustering configurations. The models were assessed using Silhouette Score, Davies-Bouldin Index, and Calinski-Harabasz Score. These evaluation metrics helped compare cluster separation, compactness, and overall clustering quality.

The resulting restaurant clusters can support business applications such as restaurant segmentation, competitor analysis, targeted marketing, and customer discovery. Restaurants with similar pricing, ratings, popularity, and cuisine diversity can be grouped together, helping businesses understand market segments and identify potential opportunities.

In conclusion, this project demonstrates how data analytics, statistical analysis, NLP, and unsupervised machine learning can be combined to extract meaningful insights from restaurant datasets. It provides practical experience in the complete data science workflow, from data collection and preprocessing to model development and evaluation. Future improvements could include advanced sentiment analysis of customer reviews, integration of real-time restaurant data, interactive dashboards, and deployment of the selected clustering model through a web application.

D. Identify outliers using IQR

````python
def detect_outliers(data, column):

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    print("Column:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))

    return outliers
````

````python
# Detect outliers in Cost
cost_outliers = detect_outliers(
    metadata_clean, 'Cost'
)

# Detect outliers in Rating
rating_outliers = detect_outliers(
    reviews_clean, 'Rating'
)

# Detect outliers in Pictures
picture_outliers = detect_outliers(
    reviews_clean, 'Pictures'
)
````

````python

# Import visualization libraries

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

sns.set_style("whitegrid")

# Create visualization copies
meta_viz = metadata_clean.copy()
reviews_viz = reviews_clean.copy()

# Ensure Cost is numerical
meta_viz['Cost'] = pd.to_numeric(
    meta_viz['Cost'], errors='coerce'
)

# Ensure Rating is numerical
reviews_viz['Rating'] = pd.to_numeric(
    reviews_viz['Rating'], errors='coerce'
)

# Create cuisine count
meta_viz['Cuisine_Count'] = (
    meta_viz['Cuisines']
    .fillna('')
    .apply(lambda x: len([c for c in x.split(',') if c.strip()]))
)

# Create restaurant-level review statistics
restaurant_stats = reviews_viz.groupby('Restaurant').agg(
    Average_Rating=('Rating', 'mean'),
    Review_Count=('Review', 'count'),
    Average_Pictures=('Pictures', 'mean')
).reset_index()

# Merge aggregated reviews with metadata for analysis
eda_df = meta_viz.merge(
    restaurant_stats,
    left_on=meta_viz['Name'].str.strip().str.casefold(),
    right_on=restaurant_stats['Restaurant'].str.strip().str.casefold(),
    how='left'
)

# Remove redundant merge keys if present
eda_df.drop(
    columns=['key_0', 'Restaurant'],
    errors='ignore',
    inplace=True
)

print("EDA preparation completed!")
print("Restaurant-level dataset:", eda_df.shape)
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

I recommend these three hypotheses because they are connected to our previous visualizations.


Hypothesis	Research question	Related chart

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀)

There is no statistically significant difference in the average customer ratings of budget restaurants and premium restaurants.

                     [H₀: µBudget = µPremium]


Alternative Hypothesis (H₁)

There is a statistically significant difference in the average customer ratings of budget restaurants and premium restaurants.

                     [H₁: µBudget ≠ µPremium]

#### 2. Perform an appropriate statistical test.

Q. Which statistical test should we use?
We will use the Independent Samples T-Test (Welch's t-test).

Why?
- We are comparing two independent groups.
- The dependent variable is numerical (average restaurant rating).
- The two groups are budget and premium restaurants.
- Welch's t-test does not require equal variances between the groups.

````python
# Perform Statistical Test to obtain P-Value

from scipy.stats import ttest_ind

# Create budget and premium restaurant groups

budget_ratings = eda_df.loc[
    eda_df['Price_Category'] == 'Budget',
    'Average_Rating'
].dropna()

premium_ratings = eda_df.loc[
    eda_df['Price_Category'] == 'Premium',
    'Average_Rating'
].dropna()

# Perform Welch's independent samples t-test

t_stat, p_value = ttest_ind(
    budget_ratings,
    premium_ratings,
    equal_var=False
)

print("Budget Restaurants:", len(budget_ratings))
print("Premium Restaurants:", len(premium_ratings))

print("Budget Average Rating:", budget_ratings.mean())
print("Premium Average Rating:", premium_ratings.mean())

print("T-statistic:", t_stat)
print("P-value:", p_value)

# Decision

alpha = 0.05

if p_value < alpha:
    print("\nReject the Null Hypothesis.")
    print("There is a significant difference in ratings.")
else:
    print("\nFail to Reject the Null Hypothesis.")
    print("There is insufficient evidence of a significant difference.")
````

##### Which statistical test have you done to obtain P-Value?

I performed the Welch's Independent Two-Sample t-test using the ttest_ind() function from the scipy.stats library to calculate the p-value. This test compares the mean average customer ratings of two independent groups: Budget restaurants and Premium restaurants.

##### Why did you choose the specific statistical test?

I chose Welch's Independent Two-Sample t-test because I wanted to determine whether there is a statistically significant difference between the mean average ratings of Budget and Premium restaurants.

This test is suitable because:
1. The two groups (Budget and Premium restaurants) are independent.
2. The variable being compared (Average Rating) is numerical.
3. Welch's t-test does not require the two groups to have equal variances.
4. It helps determine whether the observed difference in average ratings is statistically significant.
I used a significance level of 0.05 (5%) to make the final decision about the null hypothesis.

In simple terms: This test helps us understand whether Budget and Premium restaurants genuinely differ in their average customer ratings or whether the observed difference could be due to random variation.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
import numpy as np

numerical_columns = ['Cost', 'Rating', 'Pictures']

for col in numerical_columns:
    data = pd.to_numeric(
        reviews_clean[col] if col in reviews_clean.columns
        else metadata_clean[col],
        errors='coerce'
    ).dropna()

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[(data < lower_bound) | (data > upper_bound)]

    print(f"\n{col}")
    print("Number of potential outliers:", len(outliers))
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

I used the Interquartile Range (IQR) method to identify potential outliers in numerical columns such as restaurant cost, customer ratings, and number of pictures.

The IQR method calculates the lower and upper boundaries using:
- IQR = Q3 − Q1
- Lower Bound = Q1 − 1.5 × IQR
- Upper Bound = Q3 + 1.5 × IQR
Values outside these boundaries are identified as potential outliers.

I did not automatically remove these values because unusually high restaurant costs or picture counts may represent genuine observations. I preferred to identify and examine them before deciding whether treatment was necessary.

## ***7. ML Model Implementation***

### ML Model - 1

````python
# ML Model - 1 Implementation

# Fit the Algorithm

# Predict on the model
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)
import matplotlib.pyplot as plt

# Try different numbers of clusters
k_values = range(2, 11)

silhouette_scores = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(scaled_df)

    silhouette_scores.append(
        silhouette_score(scaled_df, labels)
    )

# Find the best number of clusters
best_k = list(k_values)[silhouette_scores.index(max(silhouette_scores))]

print("Best Number of Clusters:", best_k)

# Train the final K-Means model
kmeans_model = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans_model.fit_predict(scaled_df)

# Store cluster labels
feature_df['KMeans_Cluster'] = kmeans_labels

print(feature_df[['Name', 'KMeans_Cluster']].head())
````

Evaluate model performance

````python
kmeans_silhouette = silhouette_score(scaled_df, kmeans_labels)
kmeans_db = davies_bouldin_score(scaled_df, kmeans_labels)
kmeans_ch = calinski_harabasz_score(scaled_df, kmeans_labels)

print("K-Means Evaluation")
print("Silhouette Score:", kmeans_silhouette)
print("Davies-Bouldin Index:", kmeans_db)
print("Calinski-Harabasz Score:", kmeans_ch)
````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

I used the K-Means clustering algorithm, an unsupervised machine learning technique that groups similar restaurants into clusters based on their numerical features.

The algorithm calculates the distance between data points and assigns each restaurant to the nearest cluster centroid.

I evaluated the model using:
- Silhouette Score: Measures how well-separated the clusters are. Higher values are better.
- Davies-Bouldin Index: Measures similarity between clusters. Lower values are better.
- Calinski-Harabasz Score: Measures cluster separation and compactness. Higher values are generally better.
The model helps identify groups of restaurants with similar pricing, customer ratings, popularity, and cuisine variety.

````python
# Visualizing evaluation Metric Score chart
metrics_names = [
    'Silhouette Score',
    'Davies-Bouldin Index',
    'Calinski-Harabasz Score'
]

metrics_values = [
    kmeans_silhouette,
    kmeans_db,
    kmeans_ch
]

plt.figure(figsize=(9, 5))
plt.bar(metrics_names, metrics_values)
plt.title('K-Means Evaluation Metrics')
plt.ylabel('Score')
plt.xticks(rotation=15)
plt.show()
````

#### 2. Cross- Validation & Hyperparameter Tuning

For K-Means, we will use a parameter search over different cluster counts and select the one with the highest Silhouette Score. This is more appropriate than directly applying GridSearchCV, which is primarily designed for supervised learning with a scoring function and cross-validation splits.

````python
# ML Model - 1 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)

# Fit the Algorithm

# Predict on the model
from sklearn.model_selection import ParameterGrid

param_grid = {
    'n_clusters': range(2, 11),
    'init': ['k-means++', 'random'],
    'n_init': [10, 20, 30]
}

best_score = -1
best_params = None

for params in ParameterGrid(param_grid):

    model = KMeans(
        random_state=42,
        **params
    )

    labels = model.fit_predict(scaled_df)

    score = silhouette_score(scaled_df, labels)

    if score > best_score:
        best_score = score
        best_params = params

# Train tuned model
tuned_kmeans = KMeans(
    random_state=42,
    **best_params
)

tuned_kmeans_labels = tuned_kmeans.fit_predict(scaled_df)

feature_df['Tuned_KMeans_Cluster'] = tuned_kmeans_labels

print("Best Parameters:", best_params)
print("Best Silhouette Score:", best_score)
````

##### Which hyperparameter optimization technique have you used and why?

I used Grid Search (ParameterGrid) to test different combinations of K-Means hyperparameters, including the number of clusters, initialization method, and number of initializations.

I selected this technique because it systematically evaluates different parameter combinations and helps identify the configuration that produces the highest Silhouette Score.

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

Agglomerative Clustering is a hierarchical clustering algorithm. It starts by treating each restaurant as an individual cluster and gradually combines the most similar clusters until the required number of clusters is reached.

I used Agglomerative Hierarchical Clustering to group restaurants based on similarities in their numerical features.

The algorithm uses a bottom-up approach, initially treating each restaurant as a separate cluster and progressively merging similar clusters.

I evaluated its performance using Silhouette Score, Davies-Bouldin Index, and Calinski-Harabasz Score. These metrics help measure cluster separation and compactness.

````python
from sklearn.cluster import AgglomerativeClustering

# Try different cluster counts
agg_scores = []

for k in range(2, 11):

    model = AgglomerativeClustering(
        n_clusters=k,
        linkage='ward'
    )

    labels = model.fit_predict(scaled_df)

    score = silhouette_score(scaled_df, labels)

    agg_scores.append(score)

# Find the best cluster count
agg_best_k = list(range(2, 11))[agg_scores.index(max(agg_scores))]

# Train final model
agg_model = AgglomerativeClustering(
    n_clusters=agg_best_k,
    linkage='ward'
)

agg_labels = agg_model.fit_predict(scaled_df)

feature_df['Agglomerative_Cluster'] = agg_labels

print("Best Number of Clusters:", agg_best_k)
print(feature_df[['Name', 'Agglomerative_Cluster']].head())
````

Evaluate model performance

````python
agg_silhouette = silhouette_score(scaled_df, agg_labels)
agg_db = davies_bouldin_score(scaled_df, agg_labels)
agg_ch = calinski_harabasz_score(scaled_df, agg_labels)

print("Agglomerative Clustering Evaluation")
print("Silhouette Score:", agg_silhouette)
print("Davies-Bouldin Index:", agg_db)
print("Calinski-Harabasz Score:", agg_ch)
````

### ML Model - 3

````python
# ML Model - 3 Implementation
# DBSCAN stands for Density-Based Spatial Clustering of Applications with Noise.

# Unlike K-Means, DBSCAN does not require us to specify the number of clusters in advance. It groups closely located data points and identifies isolated points as noise.

# Fit the Algorithm

# Predict on the model

from sklearn.cluster import DBSCAN

# Initialize DBSCAN
dbscan_model = DBSCAN(
    eps=0.8,
    min_samples=4
)

# Fit and predict cluster labels
dbscan_labels = dbscan_model.fit_predict(scaled_df)

feature_df['DBSCAN_Cluster'] = dbscan_labels

print("Cluster Labels:", set(dbscan_labels))
print(feature_df[['Name', 'DBSCAN_Cluster']].head())

print(
    "Number of Noise Points:",
    sum(dbscan_labels == -1)
)
````

A cluster label of -1 represents a noise point that DBSCAN could not assign to a regular cluster.

Evaluate model performance

````python
# Exclude noise points for cluster-quality evaluation
dbscan_mask = dbscan_labels != -1
dbscan_valid_labels = dbscan_labels[dbscan_mask]
dbscan_valid_data = scaled_df.loc[dbscan_mask]

n_clusters = len(set(dbscan_valid_labels))

if 2 <= n_clusters < len(dbscan_valid_labels):

    dbscan_silhouette = silhouette_score(
        dbscan_valid_data,
        dbscan_valid_labels
    )

    dbscan_db = davies_bouldin_score(
        dbscan_valid_data,
        dbscan_valid_labels
    )

    dbscan_ch = calinski_harabasz_score(
        dbscan_valid_data,
        dbscan_valid_labels
    )

    print("Silhouette Score:", dbscan_silhouette)
    print("Davies-Bouldin Index:", dbscan_db)
    print("Calinski-Harabasz Score:", dbscan_ch)

else:
    print("DBSCAN did not produce enough valid clusters for evaluation.")
````

I used DBSCAN, a density-based unsupervised machine learning algorithm, to identify groups of restaurants based on the density of their data points.

Unlike K-Means, DBSCAN does not require a predefined number of clusters. It can also identify noise points that do not belong to any cluster.

I evaluated its performance using Silhouette Score, Davies-Bouldin Index, and Calinski-Harabasz Score whenever the algorithm produced at least two valid clusters.

#### 2. Cross- Validation & Hyperparameter Tuning

We will use a grid search over different eps and min_samples values. We will choose the configuration with the highest Silhouette Score, provided it produces at least two clusters after excluding noise points.

````python
# ML Model - 3 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)

# Fit the Algorithm

# Predict on the model

best_dbscan_score = -1
best_dbscan_params = None

for eps in [0.3, 0.5, 0.7, 0.8, 1.0, 1.2, 1.5]:

    for min_samples in [3, 4, 5, 6, 8]:

        model = DBSCAN(
            eps=eps,
            min_samples=min_samples
        )

        labels = model.fit_predict(scaled_df)

        # Exclude noise points
        mask = labels != -1
        valid_labels = labels[mask]
        valid_data = scaled_df.loc[mask]

        number_of_clusters = len(set(valid_labels))

        if 2 <= number_of_clusters < len(valid_labels):

            score = silhouette_score(
                valid_data,
                valid_labels
            )

            if score > best_dbscan_score:
                best_dbscan_score = score

                best_dbscan_params = {
                    'eps': eps,
                    'min_samples': min_samples
                }

if best_dbscan_params is not None:

    tuned_dbscan = DBSCAN(**best_dbscan_params)

    tuned_dbscan_labels = tuned_dbscan.fit_predict(scaled_df)

    feature_df['Tuned_DBSCAN_Cluster'] = tuned_dbscan_labels

    print("Best Parameters:", best_dbscan_params)
    print("Best Silhouette Score:", best_dbscan_score)

else:
    print("No suitable DBSCAN configuration found.")
````
